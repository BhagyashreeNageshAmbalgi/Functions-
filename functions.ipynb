Q1)  What is the difference between a function and a method in Python?


Ans: Functions vs. Methods in Python
1. Function: A standalone block of code that performs a specific task, can be called independently, and returns values.
2. Method: A function defined inside a class, bound to an instance of the class, and operates on the instance's attributes.

Key differences:

- Scope: Functions are global or module-level, while methods are part of a class.
- Binding: Functions are not bound to an object, while methods are bound to an instance of the class.
- Access: Functions don't have direct access to class attributes, while methods do.

Example:

# Function
def greet(name):
    print(f"Hello, {name}!")

# Method
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, I'm {self.name}!")

# Usage
greet("John")  # Function call
person = Person("Jane")
person.greet()  # Method call

Q2)  Explain the concept of function arguments and parameters in Python.


Ans: Function Arguments and Parameters in Python
Parameters
1. Definition: Parameters are the variables defined in a function's definition.
2. Purpose: They receive the values passed to the function when it's called.

Arguments
1. Definition: Arguments are the values passed to a function when it's called.
2. Types:
    - Positional arguments: Passed in the order defined by the function's parameters.
    - Keyword arguments: Passed using the parameter name, allowing for flexibility in order.

Example



```python
def greet(name, age):  # name and age are parameters
    print(f"Hello, {name}! You are {age} years old.")

# Positional arguments
greet("John", 30)

# Keyword arguments
greet(name="Jane", age=25)
greet(age=25, name="Jane")  # Order doesn't matter with keyword arguments
```

Q5) What are iterators in Python and how do they differ from iterables?


Ans: Iterators vs. Iterables in Python
Iterables
1. Definition: An iterable is an object that can be looped over, such as lists, tuples, dictionaries, and sets.
2. Characteristics: Iterables can be iterated over multiple times, and they don't keep track of their position.

Iterators
1. Definition: An iterator is an object that keeps track of its position and returns the next value in a sequence each time it's called.
2. Characteristics: Iterators can only be iterated over once, and they keep track of their position.

Key Differences
1. Iteration: Iterables can be iterated over multiple times, while iterators can only be iterated over once.
2. State: Iterators maintain their state (position), while iterables don't.

```python
Example

# Iterable (list)
my_list = [1, 2, 3]
for item in my_list:
    print(item)
for item in my_list:  # Can iterate multiple times
    print(item)

# Iterator
my_iterator = iter([1, 2, 3])
for item in my_iterator:
    print(item)
# Trying to iterate again won't work
for item in my_iterator:
    print(item)  # No output
```

Q6) Explain the concept of generators in Python and how they are defined.


Ans : Generators in Python
Definition
1. Generators: A generator is a special type of iterable, similar to lists or tuples, but instead of storing all values in memory, generators generate values on-the-fly.
2. Purpose: Generators are useful for creating sequences of values without having to store them all in memory at once.

Defining Generators
1. Generator Functions: Generators are defined using functions with the yield keyword instead of return.
2. Yield Statement: When a generator function is called, it returns an iterator. Each time yield is encountered, the function returns the value and pauses its execution until the next value is requested.

```python
Example

def infinite_sequence():
    num = 0
    while True:
        yield num
        num += 1

gen = infinite_sequence()
print(next(gen))  # Output: 0
print(next(gen))  # Output: 1
print(next(gen))  # Output: 2
```

Q8)  What is a lambda function in Python and when is it typically used?

Ans: Lambda Functions in Python
Definition
1. Lambda Function: A lambda function is a small, anonymous function that can take any number of arguments, but can only have one expression.
2. Syntax: lambda arguments: expression

Typical Use Cases
1. One-Time Use: Lambda functions are useful when you need a small, one-time use function.
2. Higher-Order Functions: Lambda functions are often used as arguments to higher-order functions, such as map(), filter(), and reduce().
3. Data Processing: Lambda functions can be used to process data in a concise and efficient way.

```python
Example

# Define a lambda function that adds 2 to a number
add_two = lambda x: x + 2
print(add_two(5))  # Output: 7

# Use a lambda function with map()
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x ** 2, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

Q9)  Explain the purpose and usage of the `map()` function in Python.


Ans:map() Function in Python
Purpose
1. Applying a Function: The map() function applies a given function to each item of an iterable (such as a list, tuple, or string) and returns a map object.
2. Transforming Data: map() is useful for transforming data by applying a function to each element of an iterable.

Usage
1. Syntax: map(function, iterable)
2. Function: The function to be applied to each item of the iterable.
3. Iterable: The iterable (such as a list, tuple, or string) to be processed.

```python
Example

# Define a function that squares a number
def square(x):
    return x ** 2

# Create a list of numbers
numbers = [1, 2, 3, 4, 5]

# Use map() to square each number
squared_numbers = list(map(square, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]

# Using lambda function with map()
squared_numbers = list(map(lambda x: x ** 2, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]
```

the list

```python
def sum_even_numbers(numbers):
    """
    Returns the sum of all even numbers in the input list.

    Args:
        numbers (list): A list of numbers.

    Returns:
        int: The sum of all even numbers in the list.
    """
    return sum(num for num in numbers if num % 2 == 0)
```

Q4) Write a Python function that checks if a given number is prime or not from 1 to 200

```python
def is_prime(n):
    """
    Returns True if the number is prime, False otherwise.

    Args:
        n (int): The number to check.

    Returns:
        bool: Whether the number is prime or not.
    """
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def check_primes_in_range(start, end):
    """
    Prints prime numbers in the given range.

    Args:
        start (int): The start of the range (inclusive).
        end (int): The end of the range (inclusive).
    """
    for num in range(start, end + 1):
        if is_prime(num):
            print(f"{num} is prime")
```

```python
start = 1
end = 200
print(f"Prime numbers between {start} and {end}:")
check_primes_in_range(start, end)
```

Q5) Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms

```python
class FibonacciIterator:
    """
    An iterator class that generates the Fibonacci sequence up to a specified number of terms.

    Args:
        n (int): The number of terms in the Fibonacci sequence.
    """
    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1
        self.current_term = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current_term >= self.n:
            raise StopIteration
        if self.current_term == 0:
            result = self.a
        elif self.current_term == 1:
            result = self.b
        else:
            result = self.a + self.b
            self.a, self.b = self.b, result
        self.current_term += 1
        return result

# Example usage
n = 10
fib_iterator = FibonacciIterator(n)
print(f"Fibonacci sequence up to {n} terms:")
for num in fib_iterator:
    print(num)
```

Q6) Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(n):
    """
    A generator function that yields the powers of 2 up to a given exponent.

    Args:
        n (int): The maximum exponent.

    Yields:
        int: The power of 2 for each exponent from 0 to n.
    """
    for i in range(n + 1):
        yield 2 ** i

# Example usage
n = 5
print(f"Powers of 2 up to exponent {n}:")
for power in powers_of_two(n):
    print(power)
```

Q7)  Implement a generator function that reads a file line by line and yields each line as a string.

```python
#File Reader Generator Function
#Code

def read_file_line_by_line(file_path):
    """
    A generator function that reads a file line by line and yields each line as a string.

    Args:
        file_path (str): The path to the file.

    Yields:
        str: Each line in the file.
    """
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line.strip()
    except FileNotFoundError:
        print(f"File not found: {file_path}")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
file_path = 'example.txt'
print(f"Reading file: {file_path}")
for line in read_file_line_by_line(file_path):
    print(line)
```

Q9)  Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
def celsius_to_fahrenheit(celsius):
    """
    Converts a temperature from Celsius to Fahrenheit.

    Args:
        celsius (float): The temperature in Celsius.

    Returns:
        float: The temperature in Fahrenheit.
    """
    return (celsius * 9/5) + 32

# Define a list of temperatures in Celsius
celsius_temps = [0, 10, 20, 30, 40]

# Use map() to convert the temperatures to Fahrenheit
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

# Print the temperatures in Fahrenheit
print("Celsius\tFahrenheit")
for celsius, fahrenheit in zip(celsius_temps, fahrenheit_temps):
    print(f"{celsius}\t{fahrenheit}")
```

Q10)  Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def is_vowel(char):
    """
    Checks if a character is a vowel.

    Args:
        char (str): The character to check.

    Returns:
        bool: True if the character is a vowel, False otherwise.
    """
    return char.lower() in 'aeiou'

def remove_vowels(input_string):
    """
    Removes all vowels from a given string.

    Args:
        input_string (str): The string to remove vowels from.

    Returns:
        str: The string without vowels.
    """
    return ''.join(filter(lambda x: not is_vowel(x), input_string))

# Example usage
input_string = "Hello World"
print(f"Original string: {input_string}")
print(f"String without vowels: {remove_vowels(input_string)}")
```
